```python
from scipy.stats import spearmanr

kolom_ordinal = [
      'exercise_level',
      'insurance_plan',
      'bp_category',
      'bmi_category'
]

for var in kolom_ordinal:

    # Spearman
    rho, p_spearman = spearmanr(
        df_insurance[var],
        df_insurance['annual_medical_cost_usd']
    )

    print(f"\n{var}")
    print(f"Spearman rho      : {rho:.4f}")
    print(f"Spearman p-value  : {p_spearman:.4g}")
```

```python
# Terdapat hubungan/korelasi yang signifikan secara statistik, tetapi kekuatan korelasinya lemah atau sangat lemah.
```

```python
# Uji Kruskal-Wallis

from scipy.stats import kruskal

for var in kolom_ordinal:

    # Kelompok price berdasarkan kategori
    groups = [
        group['annual_medical_cost_usd'].values
        for _, group in df_insurance.groupby(var)
    ]

    # Kruskal-Wallis
    h_stat, p_kruskal = kruskal(*groups)

    print(f"\n{var}")
    print(f"Kruskal-Wallis H  : {h_stat:.4f}")
    print(f"Kruskal p-value   : {p_kruskal:.4g}")

# Jika p-value < 0.05, lanjut post-hoc Dunn's test
# untuk mengetahui kelompok mana yang berbeda
```

# Uji Asosiasi Kategorik Nominal dgn Target

```python
from scipy.stats import mannwhitneyu, kruskal
import pandas as pd

# Variabel nominal
nominal_cols = [
    'gender', # binary
    'smoker', # binary
    'region',
    'occupation',
    'diabetes', # binary
    'marital_status'
]

hasil = []

for col in nominal_cols:

    # Jumlah kategori
    n_kategori = df_insurance[col].nunique()

    # Kelompok annual medical cost berdasarkan kategori
    groups = [
        group['annual_medical_cost_usd'].values
        for _, group in df_insurance.groupby(col)
    ]

    if n_kategori == 2:

        # Mann-Whitney U
        stat, p_value = mannwhitneyu(
            groups[0],
            groups[1],
            alternative='two-sided'
        )

        test = 'Mann-Whitney U'

    else:

        # Kruskal-Wallis
        stat, p_value = kruskal(*groups)

        test = 'Kruskal-Wallis'

    hasil.append({
        'Variabel': col,
        'Jumlah Kategori': n_kategori,
        'Uji': test,
        'Statistic': stat,
        'p-value': p_value,
        'Kesimpulan': (
            'Signifikan'
            if p_value < 0.05
            else 'Tidak signifikan'
        )
    })

hasil_nominal = pd.DataFrame(hasil)

hasil_nominal
```

```python
import pandas as pd

variabel_signifikan = [
    'smoker',
    'diabetes',
    'marital_status'
]

target = 'annual_medical_cost_usd'

ringkasan = []

for col in variabel_signifikan:
    data = df_insurance[[col, target]].dropna()

    for kategori, group in data.groupby(col, observed=True):
        nilai = group[target]

        ringkasan.append({
            'Variabel': col,
            'Kategori': kategori,
            'N': len(nilai),
            'Mean': nilai.mean(),
            'Median': nilai.median(),
            'Q1': nilai.quantile(0.25),
            'Q3': nilai.quantile(0.75),
            'IQR': nilai.quantile(0.75) - nilai.quantile(0.25)
        })

ringkasan_signifikan = pd.DataFrame(ringkasan)

ringkasan_signifikan.round(2)
```

# Asosiasi antar Kategorik baik Nominal dan Ordinal

# --> Chi-Square dan Cramér's V
berfungsi untuk menganalisis asosiasi antara dua variabel kategorikal.

Chi-Square menguji apakah terdapat asosiasi secara statistik, sedangkan Cramér's V mengukur kekuatan asosiasi tersebut.

```python
from scipy.stats import chi2_contingency
import pandas as pd

kat_cols = [
      'exercise_level',
      'insurance_plan',
      'bp_category',
      'bmi_category',
      'gender',
      'smoker',
      'region',
      'occupation',
      'diabetes',
      'marital_status'
]

hasil = []

for i in range(len(kat_cols)):
    for j in range(i + 1, len(kat_cols)):

        col1 = kat_cols[i]
        col2 = kat_cols[j]

        tabel = pd.crosstab(
            df_insurance[col1],
            df_insurance[col2]
        )

        chi2, p_value, dof, expected = chi2_contingency(tabel)

        n = tabel.values.sum()
        r, k = tabel.shape

        cramers_v = (
            chi2 / (n * min(r - 1, k - 1))
        ) ** 0.5

        hasil.append({
            'Variabel 1': col1,
            'Variabel 2': col2,
            'Chi-Square': chi2,
            'p-value': p_value,
            "Cramer's V": cramers_v
        })

hasil_chi = pd.DataFrame(hasil)

hasil_chi
```

```python
# Yang signifikan p value < 0.05:

# bp_category × bmi_category  Cramers V (0.144) asosiasi lemah

# bp_category × smoker (0.308) asosiasi sedang

# bp_category × diabetes (0.153)  asosiasi sedang

# bp_category × marital_status (0.141) asosiasi lemah

# bmi_category × diabetes (0.094) asosiasi sangat lemah

#gender x occupation (0.052) asosiasi sangat lemah

#smoker x occupation (0.051) asosiasi sangat lemah

# diabetes x marital status (asosiasi sangat lemah)
```

```python
#Pipeline Preprocessing
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, OneHotEncoder, FunctionTransformer
import numpy as np

# Biner yang sudah 0/1 -> passthrough
categoric_binary = ['smoker', 'diabetes','gender']

# Nominal yang masih string -> OHE
categoric_nominal = ['region', 'occupation', 'marital_status']


cols_numeric_no_log = [
    col for col in numeric
    if col not in cols_log
]

preprocessor = ColumnTransformer(
    transformers=[

        # Numerik yang perlu log
        (
            'numeric_log',
            Pipeline([
                ('log', FunctionTransformer(np.log1p, feature_names_out='one-to-one')),
                ('scaler', StandardScaler())
            ]),
            cols_log
        ),

        # Numerik yang tidak perlu log
        ('numeric', StandardScaler(), cols_numeric_no_log),

        # Ordinal -> sudah berupa angka
        ('ordinal', 'passthrough', categoric_ordinal),

        # Biner -> sudah 0/1
        ('binary', 'passthrough', categoric_binary),

        # Nominal -> OHE (gender cuma 2 kategori, jadi cukup 1 kolom)
        (
            'nominal',
            OneHotEncoder(
                handle_unknown='ignore',
                sparse_output=False,
            ),
            categoric_nominal
        )
    ]
)
```

```python
from sklearn.compose import TransformedTargetRegressor
from sklearn.pipeline import Pipeline
from sklearn.model_selection import KFold, cross_validate


kfold = KFold(n_splits=5, shuffle=True, random_state=42)

scoring = {
    'MAE': 'neg_mean_absolute_error',
    'MAPE': 'neg_mean_absolute_percentage_error',
    'MSE': 'neg_mean_squared_error',
    'RMSE': 'neg_root_mean_squared_error',
    'R2': 'r2'
}

def buat_pipeline(model, log_target=True):
    pipe = Pipeline([
        ('preprocessor', preprocessor),
        ('model', model)
    ])
    if log_target:
        return TransformedTargetRegressor(
            regressor=pipe,
            func=np.log1p,
            inverse_func=np.expm1
        )
    return pipe

def evaluasi(models, log_target=True):
    hasil = []
    for nama_model, model in models.items():

        # Baseline tetap polos (mean di skala asli)
        if nama_model == 'Baseline':
            pipeline = model
        else:
            pipeline = buat_pipeline(model, log_target=log_target)

        cv_result = cross_validate(
            pipeline,
            X_train,
            y_train,
            cv=kfold,
            scoring=scoring,
            n_jobs=-1,
            return_train_score=True
        )

        hasil.append({
            'Model': nama_model,
            'MAE': -cv_result['test_MAE'].mean(),
            'MAE_std': cv_result['test_MAE'].std(ddof=1),
            'MAPE': -cv_result['test_MAPE'].mean(),
            'MAPE_std': cv_result['test_MAPE'].std(ddof=1),
            'MSE': -cv_result['test_MSE'].mean(),
            'MSE_std': cv_result['test_MSE'].std(ddof=1),
            'RMSE': -cv_result['test_RMSE'].mean(),
            'RMSE_std': cv_result['test_RMSE'].std(ddof=1),
            'R2': cv_result['test_R2'].mean(),
            'R2_std': cv_result['test_R2'].std(ddof=1),
            'R2_train': cv_result['train_R2'].mean(),
            'R2_train_std': cv_result['train_R2'].std(ddof=1),
        })

    return pd.DataFrame(hasil).set_index('Model').sort_values('RMSE')
```

```python
import json
import joblib
import numpy as np
import pandas as pd

folder = '/content/drive/MyDrive/ml_statistika/insurance/'

model = joblib.load(folder + 'model_final.joblib')
with open(folder + 'metadata.json', encoding='utf-8') as f:
    meta = json.load(f)


def siapkan_input(raw: pd.DataFrame) -> pd.DataFrame:
    """Ubah data mentah (format CSV asli) jadi 24 fitur yang dipakai model."""
    df = raw.copy()
    enc = meta['encoding_manual']

    # bp_category dari string 'sys/dia'
    bp = df['blood_pressure'].astype(str).str.strip().str.split('/', expand=True)
    s = pd.to_numeric(bp[0], errors='coerce')
    d = pd.to_numeric(bp[1], errors='coerce')
    bp_label = np.select(
        [(s >= 140) | (d >= 90), (s >= 130) | (d >= 80), (s >= 120)],
        ['stage_2', 'stage_1', 'elevated'],
        default='normal'
    )
    df['bp_category'] = pd.Series(bp_label, index=df.index).map(enc['ordinal']['bp_category'])

    # bmi_category (WHO)
    bmi_label = pd.cut(
        df['bmi'],
        bins=[0, 18.5, 25, 30, float('inf')],
        labels=['underweight', 'normal', 'overweight', 'obese'],
        right=False
    ).astype(str)
    df['bmi_category'] = bmi_label.map(enc['ordinal']['bmi_category'])

    # binary + ordinal
    for col, mapping in enc['binary'].items():
        df[col] = df[col].astype(str).str.strip().map(mapping)
    for col in ['exercise_level', 'insurance_plan']:
        df[col] = df[col].astype(str).str.strip().map(enc['ordinal'][col])

    # interaksi BMI
    df['bmi_x_smoker'] = df['bmi'] * df['smoker']
    df['bmi_x_age'] = df['bmi'] * df['age']
    df['bmi_x_diabetes'] = df['bmi'] * df['diabetes']
    df['bmi_x_chronic'] = df['bmi'] * df['chronic_diseases']
    df['bmi_x_exercise'] = df['bmi'] * df['exercise_level']

    X_new = df[meta['fitur_model']]
    assert not X_new.isna().any().any(), (
        f"Ada NaN, cek label yang nggak dikenal:\n{X_new.isna().sum()[X_new.isna().sum() > 0]}"
    )
    return X_new


def prediksi(raw: pd.DataFrame) -> pd.DataFrame:
    pred = model.predict(siapkan_input(raw))
    itv = meta['interval_estimasi']
    return pd.DataFrame({
        'prediksi_usd': pred.round(0),
        'batas_bawah': (pred * itv['ratio_lower']).round(0),
        'batas_atas': (pred * itv['ratio_upper']).round(0),
    }, index=raw.index)


# ============================================================
# TEST 1: dummy dengan format data awal
# (kolom customer_id, doctor_visits, hospitalizations, target nggak perlu;
#  kalau ada, otomatis diabaikan)
# ============================================================
dummy = pd.DataFrame([
    {
        'age': 56, 'gender': 'Male', 'bmi': 25.2, 'children': 2, 'smoker': 'Yes',
        'region': 'Northeast', 'occupation': 'Driver', 'annual_income_usd': 77885,
        'exercise_level': 'Moderate', 'chronic_diseases': 0,
        'alcohol_consumption_per_week': 13, 'insurance_plan': 'Basic',
        'blood_pressure': '150/95', 'diabetes': 'Yes', 'cholesterol': 224,
        'sleep_hours': 7.6, 'stress_level': 4.5, 'marital_status': 'Married',
    },
    {
        'age': 20, 'gender': 'Male', 'bmi': 30.1, 'children': 1, 'smoker': 'No',
        'region': 'Central', 'occupation': 'Teacher', 'annual_income_usd': 85811,
        'exercise_level': 'Low', 'chronic_diseases': 1,
        'alcohol_consumption_per_week': 1, 'insurance_plan': 'Standard',
        'blood_pressure': '125/75', 'diabetes': 'No', 'cholesterol': 209,
        'sleep_hours': 7.5, 'stress_level': 5.7, 'marital_status': 'Single',
    },
])
```

```python
# Gabungkan input + hasil prediksi
hasil = prediksi(dummy)           # DataFrame 3 kolom output
tabel_final = pd.concat([dummy, hasil], axis=1)

# Tampilkan
print(tabel_final.T)              # transpose → fitur jadi baris
```
